```python
def Grid_Search(X_train, y_train,input_random_seed,currency_name,condition,tf):
        
    import numpy as np
    import pandas as pd
    from sklearn.preprocessing import StandardScaler
    from sklearn.model_selection import train_test_split
    from sklearn.metrics import confusion_matrix
    from IPython.display import clear_output
    from numpy import arange
    import datetime as dt
    from sklearn.model_selection import RandomizedSearchCV, GridSearchCV, train_test_split
    from sklearn import model_selection
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.model_selection import cross_val_score
    from sklearn.metrics import classification_report, confusion_matrix
    import pickle
    import h5py
    
    import random
    random.seed(input_random_seed) 
    
    rfc = {}
    name = currency_name+"_"+condition+"_"+tf
    
    
    
    
    param_grid = { 
    'n_estimators': [500,1000,2000],
    'max_features': [ 'sqrt', 'log2'],
    'max_depth':[5,10,15],
    'min_samples_leaf' :[2,3,4,5]  
    }

    rfc[name] = RandomForestClassifier()

    CV_rfc = GridSearchCV(estimator=rfc[name], param_grid=param_grid, cv= 5)
    CV_rfc.fit(X_train, y_train)
    best_params = pd.DataFrame()
    best_params = CV_rfc.best_params_
    
    
    
    
    return best_params
```

```python
def RF_Model(n_estimators,max_depth,max_features,min_samples_leaf,X_train,y_train,X_test,input_random_seed,currency_name,condition,tf,strategy):
        
    import numpy as np
    import pandas as pd
    from sklearn.preprocessing import StandardScaler
    from sklearn.model_selection import train_test_split
    from sklearn.metrics import confusion_matrix
    from IPython.display import clear_output
    from numpy import arange
    import datetime as dt
    from sklearn.model_selection import RandomizedSearchCV, GridSearchCV, train_test_split
    from sklearn import model_selection
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.model_selection import cross_val_score
    from sklearn.metrics import classification_report, confusion_matrix
    import pickle
    import h5py
    
    
    import eli5
    from eli5.sklearn import PermutationImportance
    
    
    
    import random
    random.seed(input_random_seed) 
    
    rfc = {}
    name = currency_name+"_"+condition+"_"+tf
    
    
    
    
    rfc[name] = RandomForestClassifier(n_estimators=n_estimators,
                                max_depth=max_depth,
                                max_features=max_features,
                                min_samples_leaf=min_samples_leaf)
    
    
    my_model = rfc[name].fit(X_train,y_train)
    
    
    
    perm = PermutationImportance(my_model, random_state=1).fit(X_train,y_train)
    permute_table = eli5.explain_weights_df(perm, feature_names = X_train.columns.tolist())
    
    
    
    filename = name+'.sav'
    pickle.dump(rfc[name], open(filename, 'wb'))



    #-----For Overalll 2019 & 2020-----#
    rfc_predict = rfc[name].predict_proba(X_test)
    rfc_predict = pd.DataFrame(rfc_predict)


    #----Training-----#
    rfc_predict_train = rfc[name].predict_proba(X_train)
    rfc_predict_train = pd.DataFrame(rfc_predict_train)
    
    
    return rfc_predict,rfc_predict_train,permute_table

```
